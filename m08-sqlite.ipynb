# Python SQLite 

- SQLite is a lightweight, serverless relational database that is built into Python using the sqlite3 module.
- Unlike files (TXT, JSON, CSV), SQLite stores data in tables and allows fast querying using SQL.
- Why do we need SQLite?
    - Files are good for storing data, but as applications grow:
        - Searching becomes slow.
        - Updating specific records becomes difficult.
        - Duplicate data increases.
        - Multiple users become harder to manage.

- SQLite solves these problems using relational databases.	


### Why SQLite?
**Advantages**<br>
✔ No installation required<br>
✔ Built into Python<br>
✔ Small and lightweight<br>
✔ Supports SQL<br>
✔ Faster than JSON for searching<br>
✔ Used in desktop and mobile applications<br>


**Importing SQLite**<br>
- *import sqlite3*<br>

sqlite3 is included with Python.<br>
No external package is required.<br>


```python
# Creating a Table


import sqlite3

connection = sqlite3.connect("company.db")
cursor = connection.cursor()

cursor.execute("""
        CREATE TABLE IF NOT EXISTS employee(
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT,
            salary REAL
        )
""")

connection.commit()
connection.close()
```

```python
# Inserting Records

connection = sqlite3.connect("company.db")
cursor = connection.cursor()

cursor.execute(
    "INSERT INTO employee (name, salary) VALUES ('Anand',50000)"
)

connection.commit()
connection.close()
```

```python
# Inserting Multiple Records

employees = [
    ("Priya",55000),
    ("Kumar",62000),
    ("Divya",58000)
]

connection = sqlite3.connect("company.db")
cursor = connection.cursor()

cursor.executemany(
"INSERT INTO employee (name, salary) VALUES (?,?)",
employees
)

connection.commit()
connection.close()
```

```python
# Reading Data

connection = sqlite3.connect("company.db")
cursor = connection.cursor()

cursor.execute("SELECT * FROM employee")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

#### fetchone()
- Returns only one record.<br>
- cursor.execute("SELECT * FROM employee")
- print(cursor.fetchone())

#### fetchmany()
- Returns limited records.
- cursor.execute("SELECT * FROM employee")
- print(cursor.fetchmany(2))




```python
# Exception Handling

import sqlite3

try:
    connection = sqlite3.connect("company.db")
    cursor = connection.cursor()
    cursor.execute("SELECT * FROM employee")
    print(cursor.fetchall())

except sqlite3.Error as e:
    print("Database Error:", e)

finally:
    connection.close()
```
