# Python Regular Expressions 

- A Regular Expression (Regex) is a sequence of characters that defines a search pattern. It is used to search, validate, extract, replace, or split text.
- Python provides built-in support for Regular Expressions through the re module.

**import re**


### Why do we need Regex?
Without regex, searching or validating text requires multiple string operations. <br>
Regex simplifies tasks like: <br>
- Email validation
- Mobile number validation
- Password validation
- Extracting dates
- Searching log files
- Parsing text files

### Basic Regex Functions

| Function | Purpose |
|---|---|
| `match()` | Checks for a match at the beginning of a string |
| `search()` | Searches for a match anywhere in a string |
| `findall()` | Returns all matching occurrences as a list |
| `finditer()` | Returns an iterator containing match objects |
| `sub()` | Replaces matching text with specified text |
| `split()` | Splits a string based on the matching pattern |
| `compile()` | Creates a reusable regular expression pattern |

```python

# re.match()
# Checks whether the pattern matches from the beginning of the string.

import re

text = "Python Programming"

result = re.match("Python", text)

print(result)
```

```python

# re.search()
# Searches anywhere inside the string.

text = "Learning Python Programming"

result = re.search("Python", text)

print(result)
```

```python

# re.findall()
# Returns all matching values as a list.

text = "Python Java Python C Python"

result = re.findall("Python", text)

print(result)
```

```python

# re.finditer()
# Returns match objects one by one.


text = "Python Java Python"

matches = re.finditer("Python", text)

for match in matches:
    print(match.start(), match.group())
```

```python

# re.split()
# Splits a string using a regex pattern.

text = "Python,Java,C++,Go"

result = re.split(",", text)

print(result)
```

```python

# re.sub()
# Replaces matching text.

text = "Python is easy"

result = re.sub("Python", "Java", text)

print(result)
```

### Common Regex Metacharacters

| Symbol | Meaning |
|---|---|
| `.` | Matches any character except a newline by default |
| `^` | Matches the start of a string |
| `$` | Matches the end of a string |
| `*` | Matches zero or more occurrences |
| `+` | Matches one or more occurrences |
| `?` | Matches zero or one occurrence |
| `[]` | Defines a character set |
| `\|` | Represents OR (alternation) |
| `()` | Groups expressions |
| `{}` | Specifies the number of repetitions |

```python
import re

text = "Python is easy. I like Python, Java and AI as choice 123"

# 1. Dot (.) - Any character
print(".", re.findall("i.", text))

# 2. Caret (^) - Beginning of string
print("^", re.findall("^Python", text))

# 3. Dollar ($) - End of string
print("$", re.findall("123$", text))
```

```python

text =  "I like Python, Pthon, Pyyython, color and colour"

# 4. Asterisk (*) - Zero or more occurrences
print("*", re.findall("Py*thon", text))

# 5. Plus (+) - One or more occurrences
print("+", re.findall("Py+thon", text))

# 6. Question Mark (?) - Zero or one occurrence
print("?", re.findall("colou?r", text))
```

```python

text = "Python is Easy. I Plan to learn Java and Python. My choices are 123, 456 and 789. ha haaa haha"

# 7. Square brackets ([]) - Character set
print("[]", re.findall("[A-Z]", text))

# 8. Pipe (|) - OR operator
print("|", re.findall("Python|Java", text))

# 9. Parentheses () - Grouping
print("()", re.findall("ha+", text))
print("()", re.findall("(ha)+", text))

# 10. Curly braces ({}) - Number of repetitions
print("{}", re.findall("[0-9]{3}", text))
```

### Character Classes

| Pattern | Meaning |
|---|---|
| `\d` | Matches a digit (0-9) |
| `\D` | Matches any non-digit character |
| `\w` | Matches a word character (letters, digits, underscore) |
| `\W` | Matches any non-word character |
| `\s` | Matches whitespace (space, tab, newline, etc.) |
| `\S` | Matches any non-whitespace character |

```python
import re

text = "Emp1102 Ajay_kumar ajay@gmail.com"

# 1. \d - Matches digits (0-9)
print("\\d", re.findall(r"\d", text))

# 2. \D - Matches non-digit characters
print("\\D", re.findall(r"\D", text))
```

```python
# 3. \w - Matches word characters
print("\\w", re.findall(r"\w", text))

# 4. \W - Matches non-word characters
print("\\W", re.findall(r"\W", text))
```

```python
# 5. \s - Matches whitespace characters
print("\\s", re.findall(r"\s", text))

# 6. \S - Matches non-whitespace characters
print("\\S", re.findall(r"\S", text))
```

```python

# Email Validation

import re

email = "anand@gmail.com"

pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'

if re.match(pattern, email):
    print("Valid Email")
else:
    print("Invalid Email")
```

```python

# Mobile Number Validation

import re

mobile = "9876543210"

pattern = r"^[6-9]\d{9}$"

if re.match(pattern, mobile):
    print("Valid Number")
else:
    print("Invalid Number")
```

```python

# Password Validation
# Rules
# •	Minimum 8 characters
# •	One uppercase letter
# •	One lowercase letter
# •	One digit
# •	One special character

import re

password = "Python@123"

pattern = r'^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@#$%^&+=!]).{8,}$'

if re.match(pattern, password):
    print("Strong Password")
else:
    print("Weak Password")

# Pattern	Meaning
# (?=...)	Positive lookahead – checks whether a condition is satisfied
# .	        Any character
# *	        Zero or more occurrences
# [a-z]	    Any lowercase letter from a to z
```

```python

# Extract Numbers

import re

text = "Order123 Price450 Qty25"

numbers = re.findall(r"\d+", text)

print(numbers)
```

```python

# Extract Words

import re

text = "Python Java C++"

words = re.findall(r"\w+", text)

print(words)
```

```python

# Using compile()
# Compiles a regex for repeated use.

import re

pattern = re.compile(r"\d+")

print(pattern.findall("100 200 300"))
```

```python

# Replace all spaces with hyphens.

import re

text = "Python Regular Expression"

result = re.sub(r"\s", "-", text)

print(result)
```
