# Python Functions

```python
# a and b are parameters.

def add(a, b):
    return a + b
```

```python
add(10,4)  #10 and 4 are arguments.
```

```python
add("Dipeswar","pal")  
```

```python
# Function Documentation (Docstring)

def prod(a, b):
    """
    Returns the product of two numbers.
    """
    return a * b

print(prod(3,4))
print(prod.__doc__)
```

```python
# Type Hints

def pow(a: int, b: int) -> int:
    return a ** b
    
```

```python
pow(5,3)
```

```python
# Default Parameters

def greet(name="Anand"):
    print("Welcome", name)

greet()
greet("Bala")
```

```python
# Keyword Arguments

def employee(name, age):
    print(name, age)

employee(age=25, name="John")
```

```python
# Positional Arguments

def employee(name, age):
    print(name, age)

employee("John", 25)
```

```python
# Arbitrary Arguments (*args)
# Used when the number of arguments is unknown.

def total(*numbers):
    print(numbers)

total(10, 20)
total(1, 2, 3, 4)
total(1,3,4,6,7,8,9,2,4,5,66,77,88)
```

```python
# Keyword Arbitrary Arguments (**kwargs)

def employee(**details):
    print(details)

employee(name="John", age=25)
```

```python
# Return Multiple Values

def calculate(a, b):
    return a + b, a - b

res1,res2 = calculate(10, 5)

print(f"Sum = {res1}, Diff = {res2}")
```

```python
# Using global

count = 0

def increment():
    global count
    count += 1

increment()

print(count)
```

### Lambda Functions
Small anonymous functions.
### Syntax
lambda arguments: expression


```python
def sqr(x):
    return x * x

print(sqr(5))
```

```python
square = lambda x: x * x

print(square(5))
```

```python
add = lambda a, b: a + b

print(add(10, 20))
```

```python
# Built-in Functions
# Python provides many built-in functions.

print(len("Python"))	# 6
print(max(10, 20, 30)) # 30
print(min(5, 8, 2))	# 2
print(sum([1, 2, 3])) 	# 6
print(abs(-15))		# 15
print(round(3.456, 2))	# 3.46
```

```python
nums = [30, 30, 5, 10, 25, 20, 15]
```

```python
print(list(filter(lambda x: x % 2 == 0, nums)))
```

```python
print(set(filter(lambda x: x % 2 == 0, nums)))
```

```python
print(dict(enumerate(filter(lambda x: x % 2 == 0, nums))))
```

```python
print(list(map(lambda x: x * 2, nums)))
```

```python
print(sorted(map(lambda x: x * 2, nums)))
```

```python
print(sorted(nums))
```

```python
print(list(reversed(nums)))
```

```python
print(any(filter(lambda x: x % 2 == 0, nums)))
```

```python
print(all(filter(lambda x: x % 2 == 0, nums)))
```
