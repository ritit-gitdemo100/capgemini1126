# Python File Handling 

- File handling is the process of creating, reading, writing, updating, and deleting files using Python.
- Unlike variables, files store data permanently.
### Why do we need it?
- Variables store data only while the program is running.
- When the application closes, the data is lost.
- Files solve this problem by storing data permanently.


### Types of Files
- Text Files store readable data.
- Binary Files store non-readable data.

|Text Files	| Binary Files |
|---|---|
|.txt |	.jpg |
|.csv |	.png |
|.json |	.pdf |
|.xml |	.mp3 |


## Opening a File
Syntax
- file = open("filename", "mode")

### File Modes

| Mode | Purpose | What happens to existing data? |
|------|---------|--------------------------------|
| `r`  | Read | Preserved. The file must exist. |
| `w`  | Write | Cleared when the file opens. |
| `a`  | Append | Preserved. New data is added at the end. |
| `x`  | Create a new file | Opening fails if the file already exists. |
| `r+` | Read and write | Preserved on open. Writing can overwrite existing data. The file must exist. |
| `w+` | Read and write | Cleared when the file opens. |
| `a+` | Read and append | Preserved. Writes always go to the end. |

```python
# Writing to a file

file = open("files/employee.txt", "w")

file.write("Abdul\n")
file.write("Priya")

# Closing a file
file.close()

print("Content written in file")
```

```python
# Using with closes file automatically

with open("files/employee.txt", "w") as file:
    file.write("Anand\n")
    file.write("Gayathri\n")

print("Content written in file")
```

```python
# Appending a file

with open("files/employee.txt", "a") as file:
    file.write("Manoj\n")
    file.write("Siva\n")

print("Appended successfully")
```

```python
# Reading from a file

with open("files/employee.txt", "r") as file:
    data = file.read()

print(data)
```

```python
# Read one line

with open("files/employee.txt", "r") as file:
    data = file.readline()

print(data)
```

```python
# Returns all lines as a list.

with open("files/employee.txt") as file:
    data = file.readlines()

print(data)
```

```python
# Reading from a file

with open("files/employee.txt", "r+") as file:
    names = file.readlines()
    for name in names:
        print(name, end="")
        
    file.write(input("Enter a name : ")+"\n")

```

```python
# Reading from a file

with open("files/employee.txt", "w+") as file:
    file.write(input("Enter a name : ")+"\n")

    print("Reading...")
    file.seek(0)
    names = file.readlines()
    for name in names:
        print(name, end="")
    
```

## Exception Handling

```python
try:
    with open("files/employee1.txt") as file:
        print(file.read())

except FileNotFoundError:
    print("File not found")

# Applications should never crash because a file is missing.
```


### Binary File Modes

| Mode | Description |
|------|-------------|
| `rb` | Read bytes from an existing file. |
| `wb` | Write bytes. Creates the file if missing or clears it when opened. |
| `ab` | Append bytes at the end. Creates the file if missing. |
| `rb+` | Read and write bytes in an existing file. Writes can overwrite data at the current position. |
| `wb+` | Read and write bytes. Creates the file if missing or clears it when opened. |
| `ab+` | Read bytes and append new bytes at the end. Creates the file if missing. |


```python
# Copying a Binary File 
# One of the most common uses of binary files is copying images, PDFs, or videos.

with open("images/logo.png", "rb") as source:
    data = source.read()

with open("images/logo-copy.png", "wb") as destination:
    destination.write(data)

print("File copied successfully")
```

```python

# This approach is memory-efficient for large files.

with open("files/demo.pdf", "rb") as source, open("files/demo1.pdf", "wb") as destination:
    while True:
        chunk = source.read(100)
        if not chunk:
            break
        destination.write(chunk)
```

## Working with JSON

```python
# Write JSON

import json

students = [
    {"name":"Anand","age":22},
    {"name":"Gayathri","age":23}
]

with open("files/student.json","w") as file:
    json.dump(students,file,indent=4)
```

```python
# Read JSON

with open("files/student.json","r") as file:
    data=json.load(file)

print(data)
```

```python

student = {"name": "Bob", "age": 22}

# Convert the dictionary to a JSON string
json_text = json.dumps(student)
print("JSON string:", json_text)
print(type(student))
print(type(json_text))

# Convert the JSON string back to a dictionary
data = json.loads(json_text)
print("Python dictionary:", data)
print("Student name:", data["name"])
```

## Working with CSV

```python
# Write CSV

import csv

with open("files/employee.csv","w",newline="") as file:
    writer=csv.writer(file)

    writer.writerow(["Name","Salary"])
    writer.writerow(["Anand",50000])
```

```python
# Read CSV

with open("files/employee.csv") as file:
    reader=csv.reader(file)

    for row in reader:
        print(row)
```
