# Assignment: Training Institute Records

Create a Jupyter notebook named `training_records.ipynb` that stores and displays records for a training institute. Choose a design that keeps your code readable, avoids unnecessary repetition, and makes it easy to add more records.

## Record requirements

| Student | Non-technical employee | Technical employee |
|---|---|---|
| `roll` | `empid` | `empid` |
| `name` | `name` | `name` |
| `phone` | `phone` | `phone` |
| `city` | `city` | `city` |
| `college` | `college` | `college` |
| `mark` | `desig` | `desig` |
| | `salary` | `salary` |
| | | `sub1` |

`desig` means job designation. `sub1` is the technical subject the employee teaches.

## Your task

1. Create one sample record of each type.
2. Display each record.
3. Display the total number of records. The total should update automatically when you add another record.

## Check your work

With the three sample records, the student should display **6 fields**, the non-technical employee **7 fields**, and the technical employee **8 fields**. The final line should read `Total records: 3`.


```python

# Common fields and record counter

class Person:
    college = "ABC Institute"
    total_records = 0

    def __init__(self, name, phone, city):
        self.name = name
        self.phone = phone
        self.city = city
        Person.total_records += 1

    def display(self):
        print("\n-----------------------")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")
        print(f"City: {self.city}")
        print(f"College: {self.college}")
```

```python

class Student(Person):
    def __init__(self, name, phone, city, roll, mark):
        super().__init__(name, phone, city)
        self.roll = roll
        self.mark = mark

    def display(self):
        super().display()
        print(f"Roll number: {self.roll}")
        print(f"Mark: {self.mark}")
```

```python

class NonTechEmployee(Person):
    def __init__(self, name, phone, city, empid, desig, salary):
        super().__init__(name, phone, city)
        self.empid = empid
        self.desig = desig
        self.salary = salary

    def display(self):
        super().display()
        print(f"Employee ID: {self.empid}")
        print(f"Designation: {self.desig}")
        print(f"Salary: {self.salary:,}")
```

```python

class TechEmployee(NonTechEmployee):
    def __init__(self, name, phone, city, empid, desig, salary, sub1):
        super().__init__(name, phone, city, empid, desig, salary)
        self.sub1 = sub1

    def display(self):
        super().display()
        print(f"Technical subject: {self.sub1}")
```

```python

s1 = Student("Anand", "9000000001", "Bengaluru", roll=101, mark=87)
s1.display()

n1 = NonTechEmployee("Meena", "9000000002", "Delhi", "E201", "Manager", 32000)
n1.display()  

t1 = TechEmployee("Ravi", "9000000003", "Chennai", "E202", "Python Trainer", 48000, "Python")
t1.display()

print(f"\nTotal records: {Person.total_records}")
```
