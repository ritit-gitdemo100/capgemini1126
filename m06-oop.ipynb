# OOP in Python

Object-Oriented Programming (OOP) is a programming approach where we organize code around objects rather than just functions and logic.
An object represents a real-world entity that has:
1.	Attributes → Data / properties (what it has)
2.	Methods → Functions / behavior (what it does)

### Real-world example
A Car object:
* Attributes
    - brand = "Toyota"
    - color = "Red"
    - speed = 100
* Methods
    - start()
    - stop()
    - accelerate()


## OOP Concepts in Python

Python OOP is based on four major principles:

| Principle | Meaning |
|---|---|
| Encapsulation | Binding data and methods together |
| Inheritance | Reusing code from existing classes |
| Polymorphism | Same interface, different behavior |
| Abstraction | Hiding internal implementation details |


```python
# Simple Class with a function
#------------------------------

class Student:
    def printStudent(self):
        print("Welcome to Student Object")
```

```python
s1 = Student()
s1.printStudent()
```

```python
# Class with attributes (Instance variables)
#-------------------------------------------

class Student:
    def setStudent(self, roll, name, mark):
        self.roll = roll
        self.name = name
        self.mark = mark
```

```python
s1 = Student()
s1.setStudent(101,'Anand',56.7)
print(s1.roll)
print(s1.name)
print(s1.mark)
```

```python
# Class with constructor and instance method
#-------------------------------------------
# self represents the current object.
# Instance methods that work with object data.
#-------------------------------------------

class Student:
    def __init__(self, roll, name, mark):
        self.roll = roll
        self.name = name
        self.mark = mark
        
    def printStudent(self):
        print(self.roll)
        print(self.name)
        print(self.mark)
```

```python
s1 = Student(101,'Anand',56.7)
# Python internally calls:
# Student.__init__(s1,101,'Anand',56.7)

s1.printStudent()
# Python internally calls:
# Student.printStudent(s1)
```

```python
# Class Variables
# Variables shared by all objects.

class Employee:
    company = "Google" # Class Variable
    
    def __init__(self, name):
        self.name = name # Instance Variable


e1 = Employee("John")
e2 = Employee("Alice")

print(e1.name, e1.company)
print(e2.name, e2.company)
```

```python
# Encapsulation
# Encapsulation means wrapping data and methods together and controlling access.

class BankAccount:

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount


account = BankAccount(1000)

account.deposit(500)

print(account.balance)
```

```python
# Private Variables
#---------------------
# Python uses _ conventions.
# Protected
    # self._salary
# Private
    # self.__salary

class Employee:

    def __init__(self):
        self.__salary = 50000


emp = Employee()
# print(emp.__salary)

# Name mangling
print(emp._Employee__salary)
```

```python
# Inheritance
# Inheritance allows one class to reuse another class's properties and methods.

class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")


dog = Dog()
dog.eat()
dog.bark()
```

```python
# Multiple Inheritance

class Father:
    def skill(self):
        print("Driving")

class Mother:
    def hobby(self):
        print("Music")

class Child(Father, Mother):
    pass


c = Child()

c.skill()
c.hobby()
```

```python
# Method Overriding
# Child class changes parent behavior.

class Animal:
    def sound(self):
        print("Some sound")

class Dog(Animal):
    def sound(self):
        print("Bark")
        super().sound()


dog = Dog()

dog.sound()
```

```python
# Ambiguity Diamond Problem
# Solved through MRO (Method Resolution Order).

class A:
    def show(self):
        print("In A")       

class B(A):
    def show(self):
        print("In B")   
        super().show() # This will invoke either A or C based on objects MRO

class C(A):
    def show(self):
        print("In C")

class D(B, C):
    def show(self):
        print("In D")
        super().show() 

obj = D()
obj.show()  
print(D.__mro__)  # printing the method resolution order

print("-"*20)
obj = B()
obj.show()
print(B.mro())  # printing the method resolution order of obj
```

```python
# Abstraction
# Abstraction means hiding implementation details and showing only necessary information.
# Python uses the abc module.

from abc import ABC, abstractmethod

# Abstract Class
class Animal(ABC):

    # Abstract Method
    @abstractmethod
    def sound(self):
        pass

    # Non-Abstract Method
    def sleep(self):
        print("Animal is sleeping")

class Dog(Animal):

    # Implementing abstract method
    def sound(self):
        print("Dog barks")


d = Dog()
d.sound()
d.sleep()
```

```python
# Static Methods
# A method that does not depend on object data.

class Math:

    @staticmethod
    def add(a, b):
        return a + b

# No object required.
print(Math.add(10,20))
```

```python
# Class Methods
# Works with class-level data.
# Uses cls.

class Employee:
    # Class Variable
    company = "Google"

    @classmethod
    def show_company(cls):
        print(cls.company)


Employee.show_company()
```

```python
class Employee:
    def __init__(self,name):
        self.name=name

    def __str__(self):
        return self.name


emp = Employee("John")
print(emp)
```

```python
# Composition
# One object contains another object.

class Engine:
    def start(self):
        print("Engine started")


class Car:
    def __init__(self):
        self.engine = Engine()


car = Car()
car.engine.start()
```
