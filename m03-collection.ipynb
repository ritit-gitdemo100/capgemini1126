# Collections

* #### List	[10, 20, 20]	=> Ordered, Mutable, Duplicates allowed
* #### Tuple	(10, 20, 20)	=> Ordered, Immutable, Duplicates allowed
* #### Set	{10, 20}	        => No sequence indexing, Mutable, No duplicates
* #### Dictionary	{"name": "Asha"}	=> Preserves insertion order, Mutable, Keys must be unique


### A list is an ordered, mutable collection that allows duplicate values.

```python
names_list = ["Ebi","Vinu","Anand","Vinu"]
print (names_list)
print (names_list[3])
names_list[3] = "Anu"
print (names_list)
```

### A tuple is an ordered, immutable collection.

```python
names_tup = ("Ebi","Vinu","Anand","Vinu")
print (names_tup)
print (names_tup[3])
# names_tup[3] = "Anu"
```

### A set stores unique values.

```python
names_set = {"Ebi","Vinu","Anand","Vinu"}
print (names_set)
# print (names_set[0])
# names_set[3] = "Anu"
```

### Dictionary stores key-value pairs.

```python
names_dict = { 
    101:"Ebi",
    102:"Vinu",
    103:"Anand",
    104:"Vinu" 
}
print (names_dict)
print (names_dict[101])
```
