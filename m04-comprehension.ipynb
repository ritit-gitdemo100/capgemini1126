# List Comprehension 

* #### List comprehension is a concise and Pythonic way to create a new list from an iterable (such as a list, tuple, string, or range) in a single line.
* #### Instead of writing multiple lines with a for loop and append(), you can often write the same logic in one expression.


### Syntax
#### new_list = [expression for item in iterable]

* #### expression: What you want to put into the new list.
* #### item: The current element from the iterable.
* #### iterable: The sequence you're looping through.

#### list = [v for k, v in emp.items() if k != "salary”)
#### print(list) # ['John', 'HR', 102, 30]


### Basic
### [expression for item in iterable]


```python
# Without List Comprehension

numbers = []

for i in range(5):
    numbers.append(i)

print(numbers)
```

```python
# Using List Comprehension

numbers = [i for i in range(5)]
print(numbers)
```

```python
# Square of Numbers

squares = [i * i for i in range(1, 6)]
print(squares)
```

### With Filter
### [expression for item in iterable if condition]


```python
# Even Numbers

even = [i+10 for i in range(1, 11) if i % 2 == 0]
print(even)
```

```python
# Multiply Every Element

numbers = [3,5,7,9]
result = [x * 10 for x in numbers]
print(result)
```

```python
# Convert to Uppercase

names = ["john", "alice", "bob"]
upper = [name.upper() for name in names]
print(upper)
```

```python
# String Length

names = ["Python", "Java", "C"]
lengths = [len(name) for name in names]
print(lengths)
```

```python
# Filter Strings

names = ["John", "Amy", "Alexander", "Bob"]
# result = [name for name in names if len(name) > 3]

result = [name for name in names if name.startswith('A')]

print(result)
```

### With if...else
### [true_value if condition else false_value for item in iterable]


```python
# Conditional Expression (if...else)

numbers = [1, 2, 3, 4, 5]
result = ["Even" if x % 2 == 0 else "Odd" for x in numbers]
print(result)
```

```python
# Nested Loops
# Traditional

pairs = []

for i in range(1, 3):
    for j in range(1, 4):
        pairs.append((i, j))

print(pairs)
```

### Nested Loops
### [expression for x in list1 for y in list2]


```python
# Nested Loop
# List Comprehension

pairs = [(i, j) for i in range(1, 3) for j in range(1, 4)]
print(pairs)
```

```python
# Remove Empty Strings

names = ["John", "", "Alice", "", "Bob"]
result = [name for name in names if name]
print(result)
```
