## Python Identifiers
*	Python is a case sensitive language
*	Python Identifier starts with a letter or (_) followed by letters, _ and digits.
*	Python accepts single ('), double (") and triple (''' or """) quotes to denote string literals,
*	The triple quotes are used to span the string across multiple lines. 
*	code block are called suites in Python, such as if, while, def, and class require a header line (:) and a suite.
*	Python's built-in id() function returns the address where the object/data is stored like id(name)


## Case Style	

* snake_case : Variables, functions, methods, modules (most common)
    * student_name, calculate_total()	
* PascalCase :Class names
    * StudentDetails, EmployeeService
* UPPER_CASE : Constants
    * MAX_SIZE, PI, DATABASE_URL
* Single leading underscore : Internal/private use (convention only)
    * _helper()
* Double leading underscore : Name mangling for class attributes
    * __balance
* Double leading and trailing	: Special (magic/dunder) methods
    * _ _init_ _, _ _str_ _, _ _len_ _
* Single trailing underscore : Avoid conflict with Python keywords
    * class_, from_

### Datatypes
*	Numeric: int, float, complex
*	Sequence: list, tuple, range
*	Text: str
*	Boolean: True, False
*	Set types: set, frozenset
*	Mapping: dict
*	NoneType: none
*	Bytes: b”Hello”

```python
# Numeric: int, float, complex

a = 10
print(a, type(a))

b = 10.5
print(b, type(b))

c = 2 + 3j
print(c, type(c))
```

```python
# Sequence: list, tuple, range

numbers = [10, 20, 30]
print(numbers, type(numbers))

values = (10, 20, 30)
print(values, type(values))

sequence = range(5)
print(sequence, type(sequence))
```

```python
# Text: str

name = "Hello"
print(name, type(name))
```

```python
# Boolean type

is_active = True
print(is_active, type(is_active))

is_completed = False
print(is_completed, type(is_completed))
```

```python
# Set types

unique_numbers = {10, 20, 30}
print(unique_numbers, type(unique_numbers))

fixed_numbers = frozenset({10, 20, 30})
print(fixed_numbers, type(fixed_numbers))
```

```python
# Mapping type

student = {"name": "Arjun", "age": 20}
print(student, type(student))
```

```python
# NoneType — use None as the value

result = None
print(result, type(result))
```

```python
# Bytes type

data = b"Hello"
print(data, type(data))
print(data, list(data))
```
