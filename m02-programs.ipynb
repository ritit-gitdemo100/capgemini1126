```python
# Digit to word

num = 2

match num:
    case 1: 
        print("One")
    case 2:
        print("Two")
    case 3:
        print("Three")
    case _:
        print ("Invalid number")
```

```python
ch = input("Char : ").strip()[0:1].lower()

match ch:
    case 'a'|'e'|'i'|'o'|'u':
        print (f" \'{ch}\' is a Vowel")
    case _:
        print (f" \'{ch}\' is not a Vowel")
```

```python
# Result Calculator
# -------------------
mark1 = 50
mark2 = 90
mark3 = 70

if mark1 >= 35 and mark2 >= 35 and mark3 >= 35 :
    result = "Pass"
else:
    result = "Fail"

avg = (mark1 + mark2 + mark3 ) / 3

print ("Result =",result)
print ("Grade = ",end="")
if result == "Fail":
    print ("Nil")
else:
    match avg:
        case a if a >= 75:
            print ("Distinction")
        case a if a >= 60:
            print ("First class")
        case b if b >= 50:
            print ("Second Class")
        case _:
            print("Nil")
    
```

```python
# Account Deactivation
# -------------------

pin_no = 1234
limit = 1

while limit <= 3:
    pin = int(input("Enter pin : "))
    if pin != pin_no:
        print ("Invalid Pin, ",3-limit,"attempts left")
        limit += 1
        continue
    else:
        print("Success! you have logged in")
        break
else:
    print("Account is blocked ")

print ("Thank you")
```
