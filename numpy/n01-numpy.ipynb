# Python NumPy

## 1. Introduction to NumPy

**NumPy (Numerical Python)** is a Python library used for **numerical computing**.

It provides support for:

- Multi-dimensional arrays
- Mathematical operations
- Statistical calculations
- Linear algebra
- Scientific computing

### Libraries Built on NumPy

NumPy is the foundation for many popular Python libraries, such as:

- **Pandas** – Data manipulation and analysis
- **Matplotlib** – Data visualization
- **Scikit-learn** – Machine learning
- **TensorFlow** – Deep learning and numerical computation

### Core Object in NumPy

The core object in NumPy is **`ndarray`**.

**ndarray** stands for **N-dimensional array**.

It is used to store and process numerical data efficiently.

---

## 2. Why Do We Need NumPy?

Python lists can store numerical data, but they are generally slower when performing mathematical operations on large datasets.

NumPy provides:

- Faster numerical computations
- Efficient memory usage
- Element-wise mathematical operations
- Support for multi-dimensional arrays
- Built-in mathematical and statistical functions

## Creating Numpy Arrays

np.array(object, dtype=None)

````python
a = np.array([10, 20, 30, 40], dtype=float)
print(a)
````

````python
# np.arange()
# Creates an array containing evenly spaced values within a specified interval.
# np.arange(start, stop, step, dtype=None)

a = np.arange(10)
print(a)
````

````python
print(np.arange(2,10,2, dtype=int))
````

````python
# np.linspace()
# Creates an array containing a specified number of evenly spaced values between two endpoints.
# np.linspace(start, stop, num=50, endpoint=True).
# num→ Number of values to generate.
# endpoint→ Whether to include the stop value.

a = np.linspace(0, 10, 6)
print(a)
````

````python
# np.zeros()
# Creates an array filled with zeros.
# np.zeros(shape, dtype=float)

a = np.zeros(5)
print(a)
````

````python
print(np.zeros((2,3)))
````

````python
# np.ones()
# Creates an array filled with ones.
# np.ones(shape, dtype=float)

a = np.ones(5)
print(a)
````

````python
# np.empty()
# Creates an array of the specified shape without initializing its values.
# np.empty(shape, dtype=float)

a = np.empty((2, 3))
print(a)

# The values displayed are not guaranteed to be zeros.
# They depend on the memory contents available to the array.
````

````python
# np.full()
# Creates an array filled with a specified value.
# np.full(shape, fill_value)
    
a = np.full((2,2), 100)
print(a)
````

````python
# np.eye()
# Creates a 2-D array with ones on a diagonal and zeros elsewhere.
# np.eye(N, M=None, k=0)
# N→ rows, M→ columns, k→ Diagonal position.

a = np.eye(3,4)
print(a)
````

````python
print(np.eye(3,4,2))
````

````python
# np.identity()
# Creates a square identity matrix.
# np.identity(n)

a = np.identity(3)
print(a)
````

## Speed : Python List Vs Numpy Array

````python
import time

pylist = list(range(1000000))

start = time.time()
for i in range(len(pylist)):
    pylist[i] = pylist[i] * 3
end = time.time()
    
print(pylist[0:10])
print("Time Taken : ",(end - start), "seconds")
````

````python
import time
narr = np.arange(1000000)

start = time.time()
narr = narr * 3 #Vector Operation
end = time.time()
    
print(narr[0:10])
print("Time Taken : ",(end - start), "seconds")
````

### Slicing — View vs Copy
This is one of the most important NumPy concepts.


````python
# In basic slicing, the result is generally a view of the original array rather than an independent copy.

a = np.array([10, 20, 30, 40, 50])
b = a[1:4]
print(b)

b[0] = 999
print(a)
````

````python
Creating an Independent Copy
Use: copy()

a = np.array([10, 20, 30, 40, 50])
b = a[1:4].copy()
b[0] = 999

print("Original:", a)
print("Copy:", b)
````

````python
# Boolean Indexing
# Boolean indexing allows us to select elements based on conditions.

b = a > 30
print(b)
````

````python
# Multiple Conditions

b = a[(a > 20) & (a < 50)]
print(b)
````

````python
b = a[(a <= 20) | (a >= 50)]
print(b)
````

````python
# Updating Values Using Boolean Indexing
# Suppose negative values should be replaced with zero.

a = np.array([10, -5, 20, -10, 30])
a[a < 0] = 0

print(a)
````

````python
# Array Operations

a=np.array([10,20,30])
b=np.array([1,2,3])

print(a+b)
````

````python
### Aggregate Functions
````

````python
arr=np.array([10,20,30])

print(np.sum(arr))
````

````python
print(np.min(arr))
````

````python
print(np.max(arr))
````

````python
# Reshaping Arrays
# Convert 1D array into 2D.
arr=np.arange(1,7)
print(arr)
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)

new_arr=arr.reshape(2,3)
print("\nAfter Reshaping\n")
print(new_arr)
print("Dimensions:", new_arr.ndim)
print("Shape:", new_arr.shape)
````

````python
# Flatten Array
# Convert multi-dimensional to one dimension.

arr=np.array([
[1,2],
[3,4]
])
print(arr)
print("\nAfter flatten \n")
print(arr.flatten())
````

````python
# Joining Arrays
# concatenate()

a=np.array([1,2,3])
b=np.array([4,5,6])

result=np.concatenate((a,b))

print(result)
````

````python
# Splitting Arrays

arr=np.array([1,2,3,4,5,6])
print(np.array_split(arr,3))
````

````python
# Matrix Operations

a=np.array([
[1,2],
[3,4]
])

b=np.array([
[5,6],
[7,8]
])

# Matrix multiplication:
print(np.dot(a,b))
````

````python
# Handling Missing Values

arr=np.array([10,np.nan,30])
print(np.isnan(arr))
````
