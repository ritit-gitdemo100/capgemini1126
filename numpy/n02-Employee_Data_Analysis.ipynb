# NumPy : Clean and Analyze Employee Data

**Today:** load a CSV, find mistakes, clean values, calculate salaries, and answer HR questions.

**Before starting:** keep `employee_performance.csv` in the **same folder** as this notebook.

## 1. Import NumPy
We give NumPy the short name `np`.

```python
import numpy as np

print("NumPy is ready!")
```

## 2. Load the CSV file
`genfromtxt` reads a CSV and converts missing numeric entries into `nan`. `skip_header=1` skips the column names.

**Try:** Run this cell, then look for `nan` values in the output.

```python
# Read the file from the notebook folder
data = np.genfromtxt("employee_performance.csv",
                     delimiter=",",
                     skip_header=1)

print(data)
```

## 3. Know the columns
The CSV columns are: **EmpID, Age, Experience, Salary, Performance, TrainingHours**. The order matters because NumPy uses numeric column positions.

```python
# Store column names so we can remember their positions
headers = ["EmpID", "Age", "Experience", "Salary", "Performance", "TrainingHours"]

print(headers)
```

## 4. Check the size of the data
`shape` tells us the number of rows and columns. `ndim` tells us how many dimensions the array has.

```python
print("Rows and columns:", data.shape)
print("Number of dimensions:", data.ndim)
print("Total values:", data.size)
```

## 5. Look at a few records
Remember: Python positions begin at **0**. `data[:5]` means the first five rows.

```python
# Display the first 5 employees
print(data[:5])
```

## 6. Select the salary column
`:` means all rows. Column index `3` is Salary. This uses slicing that you already know.

```python
# Select all rows from column 3 (Salary)
salary = data[:, 3]

print("Monthly salaries:")
print(salary)
```

## 7. What is a missing value?
An empty numeric field is represented by `np.nan`. We can test one value using `np.isnan()`.

```python
# Employee 105 has a missing age (row index 4, column index 1)
print("Age:", data[4, 1])
print("Is the age missing?", np.isnan(data[4, 1]))
```

## 8. Find missing values everywhere
`np.isnan(data)` checks each value and returns `True` if it is missing, otherwise `False`.

```python
# Build a True/False array showing missing entries
missing = np.isnan(data)

print(missing)
```

## 9. Count missing values
NumPy treats `True` as 1 and `False` as 0 when counting.

- `.sum()` counts all missing values.
- `.sum(axis=0)` counts missing values in each **column**.

We will avoid loops for now and print the columns separately.

```python
# Count missing values in the entire dataset
print("Total missing values:", missing.sum())

# Count missing values in each column
column_missing = missing.sum(axis=0)
print("Missing values by column:", column_missing)

print("Missing ages:", column_missing[1])
print("Missing salaries:", column_missing[3])
```

## 10. Work on a copy
Never overwrite the raw data during an initial cleaning exercise. `copy()` gives us a separate array to change.

```python
# Make a working copy of the original dataset
clean = data.copy()

print("Working copy created:", clean.shape)
```

## 11. Remove an exactly duplicated row
The file contains one identical employee row twice. `np.unique(..., axis=0)` keeps one copy of each complete row. **Note:** this also sorts the rows.

```python

print("Rows before:", clean.shape[0])

# finds and removes duplicate rows only if there is no nan
# clean = np.unique(clean, axis=0)

# Identify missing values
nan_mask = np.isnan(clean)

# Compare values AND their missing-value positions
comparison = np.concatenate(
    [np.where(nan_mask, 0, clean), nan_mask],
    axis=1
)

# Find the first occurrence of each unique row
_, indices = np.unique(comparison, axis=0, return_index=True)

# Keep original rows in their original order
clean = clean[np.sort(indices)]

print("Rows after:", clean.shape[0])
```

## 12. Find an invalid salary
A negative salary is not valid for our HR exercise. We use a condition (`< 0`) to select these records.

```python
# Find employees with a negative salary
bad_salary = clean[:, 3] < 0
print(bad_salary)
```

```python
print("Rows with negative salaries:")
print(clean[bad_salary])
```

## 13. Mark invalid salaries as missing
We do not guess the correct salary. First, replace invalid values with `np.nan`.

```python
# Change only negative salary values to nan
clean[clean[:, 3] < 0, 3] = np.nan

print("Missing salaries now:", np.isnan(clean[:, 3]).sum())
print(clean)
```

## 14. Find and mark invalid experience
Experience cannot be negative. The same technique works on column index `2`.

```python
# Show invalid experience values
print(clean[clean[:, 2] < 0])

# Replace invalid experience with nan
clean[clean[:, 2] < 0, 2] = np.nan
```

## 15. Find and mark invalid performance ratings
Our company uses ratings from **1 to 5**. The `|` symbol means **OR**. Parentheses are important around each comparison.

```python
# A rating is invalid if below 1 OR above 5
bad_rating = (clean[:, 4] < 1) | (clean[:, 4] > 5)

print("Invalid rating records:")
print(clean[bad_rating])

# Replace invalid ratings with nan
clean[bad_rating, 4] = np.nan

print("Invalid rating replaced:")
print(clean[bad_rating])
```

## 16. Fill missing ages with the median
The **median** is the middle value. `np.nanmedian()` ignores missing values while calculating it.

For this classroom exercise we use the median; in a real company, confirm missing employee details with HR.

```python
# Calculate the median age, ignoring nan
median_age = np.nanmedian(clean[:, 1])
print("Median age:", median_age)

# Replace only missing ages
clean[np.isnan(clean[:, 1]), 1] = median_age

print("Missing ages left:", np.isnan(clean[:, 1]).sum())
```

## 17. Fill missing experience
We repeat the same three steps: calculate median, locate missing values, replace them.

```python
# Calculate median experience
median_experience = np.nanmedian(clean[:, 2])

# Fill missing experience
clean[np.isnan(clean[:, 2]), 2] = median_experience

print("Median experience:", median_experience)
```

## 18. Fill missing salaries
The median is less affected by unusually high salaries than the mean.

```python
# Calculate median of valid salaries
median_salary = np.nanmedian(clean[:, 3])

# Fill missing salaries with that median
clean[np.isnan(clean[:, 3]), 3] = median_salary

print("Median salary:", median_salary)
print("Missing salaries left:", np.isnan(clean[:, 3]).sum())
```

## 19. Fill missing performance ratings
In this training dataset, the invalid rating became missing. We replace it with the median valid rating.

```python
# Calculate median valid rating
median_rating = np.nanmedian(clean[:, 4])

# Fill missing ratings
clean[np.isnan(clean[:, 4]), 4] = median_rating

print("Median rating:", median_rating)
```

## 20. Verify the cleaning
At this point we expect no missing numeric values. We simply print the count—no `assert` yet.

```python
# Check the whole cleaned array
print("Missing values remaining:", np.isnan(clean).sum())
print("Number of employee records:", clean.shape[0])
```

## 21. Calculate the average salary
Now we can safely use `np.mean()` because missing salaries have been filled.

```python
# Get the cleaned salary column
clean_salary = clean[:, 3]

# Calculate average monthly salary
average_salary = np.mean(clean_salary)

print("Average monthly salary: ₹", round(average_salary, 2))
```

## 22. Find employees above average salary
Use a condition to select only rows whose salary is above the average.

```python
# True for employees earning more than average
above_average = clean[:, 3] > average_salary

# Show matching employee records
print(clean[above_average])
print("Number of employees:", above_average.sum())
```

## 23. Calculate annual salary for everyone
This is **vectorization**: one operation works on the whole salary column.

```python
# Multiply all monthly salaries by 12
annual_salary = clean[:, 3] * 12

print("Annual salaries:")
print(annual_salary)
```

## 24. Calculate an 8% salary increment
`1.08` means 100% of the current salary plus an 8% increase.

```python
# Calculate new monthly salaries
new_salary = clean[:, 3] * 1.08

# Show the first five old and new salaries
print("Old salaries:", clean[:5, 3])
print("New salaries:", new_salary[:5])
```

## 25. Find high performers
A high performer has a rating of **4 or 5**. Boolean indexing lets us filter the employee records.

```python
# Select employees with performance >= 4
high_performers = clean[clean[:, 4] >= 4]

print("High performers:")
print(high_performers)
print("Total high performers:", len(high_performers))
```

## 26. Flag an unusually high salary
The dataset contains a salary of ₹250,000. For this lesson, we will **flag** salaries above ₹150,000 for review, not delete or change them.


```python
# Identify salaries above the example review threshold
needs_review = clean[:, 3] > 150000

print("Employees whose salary needs review:")
print(clean[needs_review])
```

## 27. Save the cleaned data
`np.savetxt()` writes the cleaned NumPy array to a new CSV file. The original file remains unchanged.

```python
# Save the numeric array as a CSV file
np.savetxt("employee_performance_cleaned_1.csv",
           clean,
           delimiter=",",
           header=",".join(headers),
           comments="",
           fmt="%.2f")

print("Saved: employee_performance_beginner_cleaned.csv")
```

## 28. Quick recap

| Task | NumPy technique |
|---|---|
| Read a CSV | `np.genfromtxt()` |
| Inspect data | `.shape`, slicing |
| Find missing values | `np.isnan()` |
| Count missing values | `.sum()`, `axis=0` |
| Remove identical rows | `np.unique(axis=0)` |
| Find invalid values | Boolean indexing |
| Fill missing values | `np.nanmedian()` |
| Analyze salaries | `np.mean()` |
| Calculate increments | Vectorization |
| Export cleaned data | `np.savetxt()` |
